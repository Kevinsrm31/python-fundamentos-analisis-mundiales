## Partidos de los mundiales con más goles

```python
# Seleccionamos los datos que son relevantes para la busqueda
df_matches_relevant = df_matches.get(['Year', 'Home Team Name', 'Away Team Name', 'Home Team Goals', 'Away Team Goals'])
df_matches_relevant.head()
```

```python
# Creamos una nueva columna con el total de goles del partido
df_matches_relevant['Total Goals'] = df_matches_relevant['Home Team Goals'] + df_matches_relevant['Away Team Goals']
df_matches_relevant.head()
```

```python
# Visualizacion de partidos con mas goles de los mundiales hasta el 2014
df_matches_relevant.sort_values(by='Total Goals', ascending=False).head(10)
```

## Goles promedio por partido

```python
# Nueva columna con el promedio de goles por partido
df_mundiales['Average Goals'] = df_mundiales['GoalsScored'] / df_mundiales['MatchesPlayed']
df_mundiales.head()
```

## Goleadores históricos de los mundiales

```python
df_players.head()
```

```python
# La columna Event nos indica lo que realizo cada jugador
# Necesitamos eliminar los eventos que esten con NaN que son celdas vacias
df_players_events = df_players.dropna(subset = ['Event'])
df_players_events.head()
```

```python
# Funcion que permitirá contar la cantidad de G que hay en la columna Event
def cantidad_goles(cadena):
  return cadena.count('G')
```

```python
df_players_events['Goles'] = df_players_events['Event'].apply(cantidad_goles)
df_players_events.head()
```

```python
# Seleccionamos las columnas que solo necesitamos
# Agrupamos los nombres de los jugadores
# Sumamos la cantidad goles de los jugadores
# Ordenamos por cantidad de goles de manera descendente
df_players_events.groupby('Player Name')[['Goles']].sum().sort_values(by='Goles', ascending=False).head(10)
```
